Analysis of price based on bedrooms

```python
housing.groupby("bedrooms")["price"].describe()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(data=housing, x="bedrooms", y="price")
plt.title("House Price by Number of Bedrooms")
plt.xlabel("Bedrooms")
plt.ylabel("Price")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.barplot(data=housing, x="bedrooms", y="price")
plt.title("Average House Price by Number of Bedrooms")
plt.xlabel("Number of Bedrooms")
plt.ylabel("Average Price")
plt.show()
```

The relationship between the number of bedrooms and house price was examined using a boxplot, a bar plot, and descriptive statistics. The bar plot shows that the average house price generally increases as the number of bedrooms increases. Houses with one bedroom have the lowest average price of approximately 2.71 million, while houses with five bedrooms have the highest average price of approximately 5.82 million. However, the average price for houses with six bedrooms decreases slightly to approximately 4.79 million.

The boxplot further indicates that houses with a greater number of bedrooms generally have higher median prices and greater variability in price. Houses with four and five bedrooms exhibit wider interquartile ranges and several high-value outliers, suggesting that these categories include both moderately priced and luxury properties. In contrast, one- and two-bedroom houses have relatively lower prices and less variation.

The descriptive statistics support these observations. The mean house price increases from 2.71 million for one-bedroom houses to 5.82 million for five-bedroom houses. The standard deviation also increases with the number of bedrooms, indicating that prices become more variable as the number of bedrooms increases. Although six-bedroom houses have a lower average price than five-bedroom houses, this result should be interpreted with caution because there are only two observations in this category, making the estimate less reliable.

Overall, the analysis suggests a positive relationship between the number of bedrooms and house price, indicating that houses with more bedrooms tend to command higher prices. However, bedroom count alone does not determine house price, as considerable variation exists within each bedroom category, implying that other factors such as house area, location, bathrooms, furnishing status, and additional amenities also influence the selling price.
